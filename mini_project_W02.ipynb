```python
"""
#RECORD CHECK  -  my version
===========================

Name  : Marzoug Salim - M01090910
Lane  : Cyber
Date  : 1/10/2026
"""

over_limit_count = 0
# counter initializtion to track how many record exceeded 100%

while True:
  label = input('Record Check: ')
  if label.lower() =='quit':
    break
    # tells the prompts the user for the label fiest and immediately breaks out of
  # the loop if "quit" is type in lowercase

  value = float(input('Value: '))
  limit = float (input('Limit: '))

  difference = limit-value
  percent = ((value/limit)*100)

  if (percent >100):
   status = 'OVER LIMIT'
   over_limit_count += 1
   # Adds one to the OVER LIMIT record
  elif (percent>90):
   status = 'WARNING'
  else :
   status = 'OK'


  print()
  print("=" * 34)
  print(f"  RECORD CHECK  -  {label}")
  print("=" * 34)
  print(f"Used    :{value:>11.2f}")
  print(f"Total   :{limit:>11.2f}")
  print(f"Free    :{difference:>11.2f}")
  print(f"Percent :{percent:>11.2f}%")
  print("Status  :    ", status)
# your report lines go here

  print("=" * 34)
  print()
  # After breaking the program count how many records are OVER LIMIT.
  print(f"Total records OVER LIMIT: {over_limit_count}")
```
